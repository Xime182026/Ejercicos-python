```python
mi_lista = [1 ,5 ,3, 2, 4]
print(mi_lista[4])# Imprime el elemento en la posición 4 de la lista, que es el número 5.
print(mi_lista[-1])# Imprime el último elemento de la lista, que también es el número 5.
mi_lista.append(6)
print(mi_lista)
mi_lista.sort(reverse=True)# Ordena la lista en orden descendente (de mayor a menor).
print(mi_lista)
frutas = ("manzaana", "pera", "naranja", "uva", "banana")
for fruta in frutas:# Itera sobre cada elemento en la tupla 'frutas' y lo asigna a la variable 'fruta' en cada iteración.
    print(fruta)

for i, fruta in enumerate(frutas):# Utiliza la función 'enumerate' para obtener tanto el índice como el valor de cada elemento en la tupla 'frutas'.
    print(f"{i}: {fruta}")# Imprime el índice y el valor de cada fruta en formato "índice: fruta".

mayusculas=[fruta.upper() for fruta in frutas]# Crea una nueva lista 'mayusculas' que contiene las frutas en mayúsculas utilizando una comprensión de listas.
print(mayusculas)
```
